```python
class Flight:
    def __init__(self, flight_id, departure, destination, price):
        self.flight_id = flight_id
        self.departure = departure
        self.destination = destination
        self.price = price


flights = [
    Flight(101, "Karachi", "Dubai", 45000),
    Flight(102, "Islamabad", "Dubai", 40000),
    Flight(103, "Karachi", "USA", 35000),
    Flight(104, "Lahore", "Dubai", 50000),
    Flight(105, "Karachi", "Istanbul", 70000),
    Flight(106, "Dubai", "Istanbul", 70000)
]


print("Travel Agent")

departure = input("Enter departure city: ").strip()
destination = input("Enter destination city: ").strip()

found_flights = []

for flight in flights:
    if (
        flight.departure.lower() == departure.lower()
        and flight.destination.lower() == destination.lower()
    ):
        found_flights.append(flight)


print("\nAvailable Flights")

if len(found_flights) == 0:
    print("No flights available")

else:
    for flight in found_flights:
        print(
            "Flight ID:", flight.flight_id,
            "| From:", flight.departure,
            "| To:", flight.destination,
            "| Price:", flight.price
        )

    cheapest_flight = found_flights[0]

    for flight in found_flights:
        if flight.price < cheapest_flight.price:
            cheapest_flight = flight

    print("\nRecommended Flight")
    print("Flight ID:", cheapest_flight.flight_id)
    print("Price:", cheapest_flight.price)

    choice = int(input("\nEnter flight ID to book: "))

    booked = False

    for flight in found_flights:
        if flight.flight_id == choice:
            booked = True

            print("\nFlight booked successfully")
            print("Flight ID:", flight.flight_id)
            print("Departure:", flight.departure)
            print("Destination:", flight.destination)
            print("Price:", flight.price)

            break

    if booked == False:
        print("Invalid Flight ID")
```
